# 1. Project Overview

## 1.1 Context and Core Problem

Lending decisions require a balance between **credit risk and economic return**.

A loan with a higher interest rate may generate greater revenue, but it may also expose the lender to a higher probability of default and greater potential loss. Conversely, restricting lending to borrowers with very low risk may reduce losses but can also limit lending volume and potential returns.

This project uses historical Lending Club loan data from 2007 to 2020 Q1 to develop an **end-to-end credit risk analytics and decisioning framework**.

The framework connects four key components:

1. **Probability of Default (PD)**
2. **Expected Loss (EL)**
3. **Expected Economic Value**
4. **Customer and Portfolio-Level Decision Analysis**

The central business question is:

> **Given a borrower's characteristics and a proposed loan structure, how can we estimate the associated credit risk and determine whether the lending decision is economically attractive?**

The project therefore moves beyond default classification toward a broader decision framework:

```text
Risk Prediction
      ↓
Probability of Default
      ↓
Expected Loss
      ↓
Expected Economic Value
      ↓
Customer-Level Decisioning
      ↓
Portfolio Strategy
```

## 1.2 Project Objectives

### Primary Objective

Develop an **interpretable and well-calibrated Probability of Default (PD) model** that estimates the likelihood of loan default using information available at or before loan origination.

For loan \(i\), the model estimates:

$$
PD_i = P(Y_i = 1 \mid X_i)
$$

where:

* \(PD_i\) is the predicted probability of default for loan \(i\)
* \(Y_i\) is the default outcome
* \(X_i\) represents borrower and loan characteristics available at the prediction point

### Secondary Objectives

#### 1. Risk Driver Analysis

Identify and quantify the borrower, credit, loan, and temporal characteristics associated with default risk.

The analysis will distinguish between:

* statistical association
* predictive importance
* model contribution
* business interpretation

#### 2. Expected Loss and Economic Value

Translate predicted default probability into an estimate of expected credit loss.

The general Expected Loss framework is:

$$
EL_i = PD_i \times LGD_i \times EAD_i
$$

where:

* \(PD_i\) = Probability of Default
* \(LGD_i\) = Loss Given Default
* \(EAD_i\) = Exposure at Default

Expected economic value can then be expressed conceptually as:

$$
EV_i = E[Revenue_i] - E[Loss_i] - E[Cost_i]
$$

The exact implementation will depend on the availability and reliability of revenue, recovery, and exposure information in the dataset.

#### 3. Customer-Level Decision Simulation

Develop a scenario-based framework that evaluates alternative loan configurations for a borrower.

The simulation will investigate how changes in:

* loan amount
* interest rate
* loan term

affect:

$$
\text{Loan Configuration}
\rightarrow
PD
\rightarrow
EL
\rightarrow
Expected\ Revenue
\rightarrow
Expected\ Value
$$

The objective is to identify a **risk-adjusted economic sweet spot** rather than simply selecting the lowest-risk loan.

#### 4. Portfolio Analysis

Analyze how portfolio characteristics and credit risk evolve over time.

The analysis will examine:

* loan volume
* borrower composition
* loan characteristics
* pricing
* default rate
* predicted risk
* expected loss
* economic performance
* vintage stability

#### 5. Portfolio Strategy

Explore how lending decisions could be allocated across borrower and risk segments to maximize expected economic value subject to defined risk constraints.

Conceptually:

$$
\max_{\text{Portfolio}}
\quad
E[Profit]
$$

subject to:

$$
E[Loss] \leq L_{max}
$$

and potentially:

$$
Capital\ Exposure \leq Budget
$$

with additional constraints introduced if supported by the data.

### Prediction Point

The PD model will use information that would reasonably be available **at or before loan origination**.

Variables generated after origination will be treated as potential target leakage. Examples include:

* subsequent payments
* recoveries
* outstanding principal
* payment dates
* post-origination performance indicators

These variables may be useful for retrospective loss analysis, but they should not be used as predictors when estimating the default probability at origination.

This distinction is fundamental to maintaining a realistic credit-risk modeling framework.

## 1.4 Analytical Framework

The project follows a progression from **risk prediction to economic decision-making**.

### Step 1. Probability of Default

Estimate the probability that a loan will default:

$$
PD_i = P(Default_i = 1 \mid X_i)
$$

### Step 2. Expected Loss

Translate default probability into expected credit loss:

$$
EL_i = PD_i \times LGD_i \times EAD_i
$$

where:

$$
LGD_i = 1 - RecoveryRate_i
$$

when recovery information is available and sufficiently reliable.

### Step 3. Expected Economic Value

Estimate the economic value of the loan:

$$
EV_i =
E[Revenue_i]
-
E[Loss_i]
-
E[Cost_i]
$$

A simplified formulation can be represented as:

$$
EV_i =
E[Interest_i + Fee_i]
-
EL_i
-
E[OperatingCost_i]
$$

The implementation will use only components that can be reasonably estimated from the available data.

### Step 4. Customer-Level Decisioning

For a given borrower, evaluate alternative loan configurations:

$$
S_i = (Amount_i, Rate_i, Term_i)
$$

For each scenario \(S_i\):

$$
S_i
\rightarrow
PD_i
\rightarrow
EL_i
\rightarrow
EV_i
$$

This allows the analysis to identify configurations that provide attractive risk-adjusted economic value.

### Step 5. Portfolio-Level Strategy

Aggregate individual loan decisions to evaluate portfolio outcomes:

$$
Portfolio\ Value
=
\sum_{i=1}^{N} EV_i
$$

while monitoring:

$$
Portfolio\ Loss
=
\sum_{i=1}^{N} EL_i
$$

This creates a connection between individual credit decisions and overall portfolio performance.

## 1.5 Analytical Questions

The project is designed to answer five major questions.

### Risk

**1. Which borrower, credit, loan, and temporal characteristics are associated with higher default risk?**

### Prediction

**2. Can these characteristics be used to produce accurate and well-calibrated Probability of Default estimates?**

### Economics

**3. How does predicted credit risk translate into Expected Loss and expected economic value?**

### Customer Decisioning

**4. For a given borrower, which combination of loan amount, interest rate, and term provides the most attractive risk-adjusted economic value?**

### Portfolio

**5. How do portfolio risk, composition, pricing, and economic performance evolve across loan vintages, and what implications does this have for lending strategy?**

## 1.8 Success Criteria

The project will not evaluate model quality based solely on classification accuracy.

Performance will be assessed across four dimensions.

### Predictive Performance

Evaluate the model's ability to discriminate between default and non-default loans using metrics such as:

* ROC-AUC
* PR-AUC
* precision
* recall
* F1 score
* KS-Stat

### Probability Quality

Because PD is used as an input to Expected Loss, the quality of the predicted probabilities is critical.

Evaluate:

* calibration curve
* Brier score
* calibration by risk band
* observed versus predicted default rate

The objective is not only to rank borrowers correctly, but also to produce probabilities that are reasonably aligned with observed outcomes.

### Economic Performance

Evaluate the implications of the model through:

* Expected Loss
* expected revenue
* expected economic value
* risk-adjusted return
* threshold sensitivity

### Robustness

Assess whether the model remains reliable under realistic changes in the data through:

* temporal validation
* vintage analysis
* feature stability
* segment-level performance
* leakage prevention

## 1.9 Project Narrative

This project develops an **end-to-end credit risk analytics and decisioning framework** using historical Lending Club loan data from 2007 to 2020 Q1.

Rather than treating default prediction as an isolated binary classification problem, the project connects **Probability of Default with Expected Loss and economic value**. This allows predictive outputs to be translated into practical lending decisions at both the individual loan and portfolio levels.

The analysis begins by identifying borrower, credit, loan, and temporal characteristics associated with default. These findings inform feature engineering and the development of a calibrated Probability of Default model using information available at or before loan origination.

The predicted PD is then incorporated into an Expected Loss and economic value framework. At the customer level, a scenario-based simulator evaluates how changes in loan amount, interest rate, and term influence the trade-off between risk and return.

At the portfolio level, the analysis examines how borrower composition, loan characteristics, pricing, risk, and economic performance evolve across vintages. Temporal validation is used to evaluate whether the resulting risk model can generalize to future periods.

The ultimate objective is to demonstrate how **statistical analysis, machine learning, and financial reasoning can be combined to move from predicting credit risk toward economically informed lending decisions**.
